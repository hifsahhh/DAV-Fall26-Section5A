```python
import pandas as pd
import numpy as np

df = pd.read_csv('adult.csv')
df.head()
```

```python
df.info()
print(df.isnull().sum())

df.replace('?', np.nan, inplace=True)

print(df.isnull().sum())

```

```python
missing_count = df.isnull().sum()
missing_percent = df.isnull().mean() * 100

missing_summary = pd.DataFrame({
    'missing_count': missing_count,
    'missing_percent': missing_percent
}).sort_values(by='missing_count', ascending=False)

print(missing_summary[missing_summary['missing_count'] > 0])
#occupation and workclasshave the most missing data
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
print("Missing values in numeric columns:")
print(df[numeric_cols].isnull().sum())

df_clean = df.copy()

df_clean['workclass'] = df_clean['workclass'].fillna('Unknown')
df_clean['occupation'] = df_clean['occupation'].fillna('Unknown')


mode_country = df_clean['native.country'].mode()[0]
print("\nMode of native.country:", mode_country)

df_clean['native.country'] = df_clean['native.country'].fillna(mode_country)

print("\nMissing values after imputation:")
print(df_clean.isnull().sum())

print("\nTotal missing values remaining:", df_clean.isnull().sum().sum())
```

```python

print(df.duplicated().sum())


cols_no_target = df.columns.drop('income')
print(df.duplicated(subset=cols_no_target).sum())

df = df.drop_duplicates()
print(df.shape)
```

```python
for col in ['education', 'marital.status', 'native.country']:
    print(col, df[col].unique())


for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

```python
print(df.describe())
print(df.describe(include='object'))

print("Age mean:", df['age'].mean(), " median:", df['age'].median())
print("Hours/week mean:", df['hours.per.week'].mean(), " median:", df['hours.per.week'].median())

#Age: mean 38.58, median 37 — slightly skewed higher. Hours-per-week: mean 40.44, median 40 — nearly symmetric, no meaningful skew.
```

```python

print(df['occupation'].value_counts().head(1))

print(df['sex'].value_counts(normalize=True) * 100)

print(df['income'].value_counts(normalize=True) * 100)
#income classes are imbalannced with a rough percentgeof 75 and 25 percent
```

```python
edu_check = df.groupby('education')['education.num'].unique()
print(edu_check)

inconsistent = edu_check[edu_check.apply(len) > 1]
print(inconsistent)
#no inconsisitencies
```

```python
import sqlite3

conn = sqlite3.connect('adult_income.db')
df.to_sql('adult_income', conn, if_exists='replace', index=False)

query = "SELECT * FROM adult_income WHERE age > 30;"
df_sql = pd.read_sql_query(query, conn)

print("Original shape:", df.shape)
print("SQL result shape:", df_sql.shape)

pandas_check = df[df['age'] > 30].shape[0]
print(pandas_check == df_sql.shape[0])

conn.close()
```

- Dataset: ~32.5K rows, 15 columns, predicts income <=50K or >50K.
- Issues found: hidden missing values (as "?"), duplicate rows, extra whitespace in text.
- Key finding: income classes are imbalanced (~3:1).
- Fixes: "?" ->NaN, missing categories -> "Unknown", duplicates dropped, whitespace stripped.
